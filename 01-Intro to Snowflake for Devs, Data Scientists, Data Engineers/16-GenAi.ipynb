# Section : GenAI with Snowflake
## Snowflake GenAI Overview - Part II

```
%%sql -r dataframe_1
CREATE OR REPLACE CORTEX SEARCH SERVICE TASTY_BYTES.ANALYTICS.TASTY_BYTES_MENU_SEARCH
    ON MENU_ITEM_NAME
    WAREHOUSE = COMPUTE_WH
    TARGET_LAG = '1 HOUR'
AS 
SELECT MENU_ITEM_NAME, TRUCK_BRAND_NAME, ITEM_CATEGORY
 FROM TASTY_BYTES.RAW_POS.MENU;
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE('mistral-7b', 'Cuales son las 3 razones por las cuales Snowflake esta posicionada como la plataforma de datos que se debe usar ?')
```

```
SELECT SNOWFLAKE.CORTEX.SUMMARIZE(
        SNOWFLAKE.CORTEX.COMPLETE('mistral-7b', 'Cuales son las 3 razones por las cuales Snowflake esta posicionada como la plataforma de datos que se debe usar ?')
        )
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE('mistral-7b', 
        CONCAT ('Dime por que esta comida es deliciosa ', MENU_ITEM_NAME)
        ) from TASTY_BYTES.RAW_POS.MENU LIMIT 5;
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE(
    'mistral-7b',
    CONCAT(
        'Responde únicamente en español. ',
        'Explica de manera breve y atractiva por qué este plato es delicioso. ',
        'No respondas en inglés. Plato: ',
        MENU_ITEM_NAME
    )
) AS DESCRIPCION
FROM TASTY_BYTES.RAW_POS.MENU
LIMIT 5;
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE(
    'mistral-7b',
    [
        {
            'role'      : 'system', 
            'content'   : 'Analyze this Snowflake review and determine the overall sentiment. Answer with just \"Possitive" ,\"Negative",or \"Neutral"' 
        },
        {
            'role'      : 'user', 
            'content'   : 'I love Snowflake because it is so simple to use."' 
        }
    ],
    {}
) as response;
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE(
    'mistral-7b',
    [
        {
            'role'      : 'system', 
            'content'   : 'Analyze this Snowflake review and determine the overall sentiment. Answer with just \"Possitive" ,\"Negative",or \"Neutral"' 
        },
        {
            'role'      : 'user', 
            'content'   : 'I love Snowflake because it is so simple to use."' 
        },
        {
            'role'      : 'assistant', 
            'content'   : 'Positive. The review expresses a positive sentiment towards Snowflake, specifically mentioning that it is \"so simple to use.\"' 
        },
        {
            'role'      : 'user', 
            'content'   : 'Based in other information you know about Snowflake, explain why the reviewer might feel they way they do"' 
        }
    ],
    {}
) as response;
```

```
SELECT SNOWFLAKE.CORTEX.COMPLETE('llama3.1-8b', 'What kind of literature was Marianne Moore known for?')
```

```
SELECT menu_item_name,SNOWFLAKE.CORTEX.AI_COMPLETE('llama3.1-8b',CONCAT('Describe this food item in one sentence: ', menu_item_name))
FROM tasty_bytes.raw_pos.menu;
```
